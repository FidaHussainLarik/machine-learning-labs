# Univariate Exploratory Data Analysis on Titanic Dataset

```python
import pandas as pd
import seaborn as sns
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('train.csv')
df.shape
df.info()
```

```python
df.sample(5)
```

```python
df.head()
```

# 1. EDA on Catagorical Columns




## Check how many survived in Titanic

```python
sns.countplot(x='Survived', data=df)
df['Survived'].value_counts()
```

# How many people were in each class


# a. CounPlot

```python
sns.countplot(x='Pclass', data=df)

```

# b. Pie chart

```python
df['Pclass'].value_counts().plot(kind='pie',autopct = '%2f')
```

```python
df['Sex'].value_counts().plot(kind='pie',autopct ='%.2f' )
```

# 2. Numerical Data

## a. Histogram

```python
import matplotlib.pyplot as plt
plt.hist(df['Age'])
```

## b. Distplot

```python
sns.distplot(df['Age'])
```

## c. BoxPlot

```python
sns.boxplot(x=df['Age'])
```

## Min and max ages

```python
min_age = df['Age'].min()
max_age = df['Age'].max()


print("Minimum age of people in Titanic: ",min_age)
print("Maximum age of people in Titanic: ",max_age)
```

```python
print("Mean age: ",df['Age'].mean())
```

```python
# Skewness of data: if value == 0, the destribution is normal , if not distribution is either left skewed or right skewed
print('Skewness: ',df['Age'].skew())
```
