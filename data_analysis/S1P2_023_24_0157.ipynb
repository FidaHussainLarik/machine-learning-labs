```python

# Question no: 02

import matplotlib.pyplot as plt

#a

print("Q2 MonthlyRent summary:")
print(df["MonthlyRent"].describe())

df["MonthlyRent"].hist(bins=20)

plt.title("Distribution of Monthly Rent")
plt.xlabel("Monthly Rent (PKR thousand)")
plt.ylabel("Frequency")
plt.show()

```

```python
#b

print("\nQ2 mean rent by Furnishing:")
print(df.groupby("Furnishing")["MonthlyRent"].mean())

# Ans: Furnished houses are expected to have the highest mean rent.
# Compare the mean and median to determine whether the mean is close
# to the median.
print("\nMedian MonthlyRent:", df["MonthlyRent"].median())
```

```python

# Question no: 03

#a

print("Q3 dtype before:", df["AreaSqft"].dtype)

df["AreaSqft"] = pd.to_numeric(df["AreaSqft"], errors="coerce")

print("Q3 dtype after:", df["AreaSqft"].dtype)
print("Missing after conversion:", df["AreaSqft"].isna().sum())


#b

df["AreaSqft"] = df["AreaSqft"].fillna(df["AreaSqft"].median())

df["Sector"] = df["Sector"].fillna(df["Sector"].mode()[0])

df = df.drop_duplicates()

print("Q3 shape after cleaning:", df.shape)
print("Total missing:", df.isna().sum().sum())

# Ans: Median is used for AreaSqFt because it is less affected
# by outliers. Mode is used for Sector because it is categorical.
```
